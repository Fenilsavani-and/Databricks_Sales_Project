## Read From CSV and Write Bronze Table

#### Drop if Table is not null.

```python
for item in INGESTION_CONFIG:
    table_name = f"bronze.{item['table']}"
    spark.sql(f"DROP TABLE IF EXISTS {table_name}")
```

```python
for item in INGESTION_CONFIG:
    print(f"Ingesting {item['source']} → bronze.{item['table']}")

    df = (
        spark.read
             .option("header", "true")
             .option("inferSchema", "true")
             .csv(item["path"])
    )

    (
        df.write
          .mode("overwrite")
          .format("delta")
          .saveAsTable(f"bronze.{item['table']}")
    )
```
